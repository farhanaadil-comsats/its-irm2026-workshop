# Start Here — Edge Intelligence, Bio-Inspired Optimization & UAV-Assisted VANETs

### A self-guided hands-on workshop · ITS IRM 2026 · Dr. Farhan Aadil

This workshop is designed so **you drive it yourself**. Each notebook below is self-contained,
runs entirely in your browser via Google Colab, and checks your own work automatically — you do not
need to wait for anyone to confirm whether your answer is correct.

## How to use these notebooks
1. Open each notebook in **Google Colab**: `File → Open notebook → Upload` (or open directly from
   GitHub/Drive if your facilitator shared a link — click "Open in Colab" if you see that button).
2. Work top to bottom. Every exercise has a **TODO cell** (where you write code) followed by a
   **self-check cell** (which tells you `PASSED` or explains exactly what's wrong).
3. If a hint is available, it's in a collapsed comment directly under the TODO cell.
4. No local installation is required — Colab provides Python, NumPy, and Matplotlib already. Notebook 3
   additionally installs the `openai` package automatically (used to call the DeepSeek API).

completely normal point to flag for facilitator support rather than a sign you're doing something wrong.
